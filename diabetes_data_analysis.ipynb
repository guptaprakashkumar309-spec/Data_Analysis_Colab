**Modification of Data**

```python
#Add a new column
df['BMI_CATEGORY']=pd.cut(df['BMI'],bins=[0,18.5,24.9,29.9,100],labels=['Underweight','Normal','Overweight','Obese'])
```

```python
df.head()
```

```python
#Replace invalid zero
df['Insulin']=df['Insulin'].replace(0,df['Insulin'].mean())
```
